```python
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader("/content/esp32_datasheet_en.pdf")
documents = loader.load()

print(len(documents))
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

textSplitter = RecursiveCharacterTextSplitter(
    chunk_size=200,
    chunk_overlap=20
)

finalDocuments = textSplitter.split_documents(documents)

print(len(finalDocuments))
print(finalDocuments[0].page_content)
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model_name="all-MiniLM-L6-v2"
)
```

```python
from langchain_community.vectorstores import FAISS

db = FAISS.from_documents(finalDocuments, embeddings)

print("Vector database created successfully!")
```

```python
query = input("Ask a question about the ESP32 datasheet: ")

docs = db.similarity_search(query)

for i, doc in enumerate(docs, 1):
    print(f"\nResult {i}")
    print(doc.page_content)
    print("-" * 50)
```
