```python
#data loading first step
from langchain_community.document_loaders import TextLoader
loader= TextLoader("/content/cppData.txt",encoding="utf-8")
txtData = loader.load()
print(txtData[0])
```

```python
#splitting data into documents
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter = RecursiveCharacterTextSplitter(chunk_size=200, chunk_overlap=20)
finalDocuments = text_splitter.split_documents(txtData)
print(finalDocuments[0])
```

```python
#This phase is embedding phase here we will convert  the textual data into vectors
from langchain_huggingface import HuggingFaceEmbeddings
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
```

```python
#vector store in database (DB)
from langchain_community.vectorstores import FAISS
db = FAISS.from_documents(finalDocuments, embeddings)
db
```

```python
query = input ("Enter your question for c++ :")
docs = db.similarity_search(query)
for i in range(len(docs)):
  print(docs[i].page_content)
```
